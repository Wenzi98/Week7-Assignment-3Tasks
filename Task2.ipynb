```python
# Import libraries
import pandas as pd
import numpy as np

# --- Create/load dataset with guaranteed df creation ---
def load_or_create_data():
    try:
        # Attempt to load existing data
        df = pd.read_csv("sales_data.csv")
        print("Dataset loaded successfully")
        return df
    except FileNotFoundError:
        print("Creating sample dataset...")
        np.random.seed(42)
        
        # Generate sample data
        data = {
            'Order_Date': pd.date_range(start='2023-01-01', end='2023-12-31', periods=100),
            'Product_Category': np.random.choice(['Electronics', 'Clothing', 'Home Goods', 'Books'], 100),
            'Unit_Price': np.round(np.random.uniform(10, 500, 100), 2),
            'Quantity': np.random.randint(1, 10, 100),
            'Region': np.random.choice(['West', 'East', 'North', 'South'], 100)
        }
        
        # Calculate derived fields
        data['Revenue'] = np.round(data['Unit_Price'] * data['Quantity'], 2)
        data['Profit'] = np.round(data['Revenue'] * np.random.uniform(0.1, 0.4, 100), 2)
        
        df = pd.DataFrame(data)
        df.to_csv("sales_data.csv", index=False)
        print("Sample dataset created")
        return df

# Guaranteed dataframe creation
df = load_or_create_data()

# --- Task 2: Basic Data Analysis ---
# 1. Basic statistics
print("\n=== Basic Statistics ===")
print(df[['Unit_Price', 'Quantity', 'Revenue', 'Profit']].describe())

# 2. Grouped analysis
print("\n=== Category Analysis ===")
category_stats = df.groupby('Product_Category').agg({
    'Revenue': ['mean', 'median'],
    'Profit': 'sum'
})
print(category_stats)

# 3. Interesting patterns
print("\n=== Key Findings ===")
# Price vs Quantity relationship
correlation = df['Unit_Price'].corr(df['Quantity'])
print(f"1. Price-Quantity Correlation: {correlation:.2f} (Negative = Higher prices sell fewer units)")

# Regional performance
region_profit = df.groupby('Region')['Profit'].sum().sort_values(ascending=False)
print(f"2. Regional Profit Ranking: {region_profit.idxmax()} region leads with ${region_profit.max():,.2f}")

# Best month analysis
df['Month'] = df['Order_Date'].dt.month_name()
monthly_sales = df.groupby('Month')['Revenue'].sum().idxmax()
print(f"3. Best Sales Month: {monthly_sales}")
```
